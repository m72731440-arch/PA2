```python
2
num_str = input("Digite um número inteiro positivo: ")

try:
  num = int(num_str)
  if num <= 0:
    print("Por favor, digite um número inteiro positivo.")
  else:
    total_sum = 0
    sum_parts = []
    for i in range(1, num + 1):
      total_sum += i
      sum_parts.append(str(i))

    print(f"Resultado: {total_sum}")
    print(f"Pois:\n{' + '.join(sum_parts)} = {total_sum}")
except ValueError:
  print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
4
num_str = input("Digite um número inteiro positivo: ")

try:
  num = int(num_str)
  if num < 0:
    print("Por favor, digite um número inteiro positivo.")
  elif num == 0:
    print("Fatorial = 1")
    print("Pois:\n0! = 1 (por definição)")
  else:
    factorial = 1
    calculation_parts = []
    for i in range(num, 0, -1):
      factorial *= i
      calculation_parts.append(str(i))

    print(f"Fatorial = {factorial}")
    print(f"Pois:\n{num}! = {' × '.join(calculation_parts)}")
except ValueError:
  print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
5
valor_saque_str = input("Digite o valor do saque: ")

try:
  valor_saque = int(valor_saque_str)

  if valor_saque <= 0:
    print("Por favor, digite um valor de saque positivo.")
  else:
    print("\nSaída esperada:")

    # Notas disponíveis (em ordem decrescente)
    nota_100 = 0
    nota_50 = 0
    nota_20 = 0
    nota_10 = 0
    nota_5 = 0
    nota_2 = 0
    nota_1 = 0

    # Usando laços (implicitamente aqui com repetidas subtrações/divisões)
    # Mas para seguir o requisito de 'laços de repetição' de forma mais explícita,
    # faremos com um loop while para cada nota ou uma estrutura if-elif para cada nota.

    # Abordagem com if/elif (sem loops explícitos para cada nota, mas o conceito é o mesmo de distribuição)
    # Vamos usar a lógica de divisão e resto para simular a iteração.

    if valor_saque >= 100:
      nota_100 = valor_saque // 100
      valor_saque %= 100

    if valor_saque >= 50:
      nota_50 = valor_saque // 50
      valor_saque %= 50

    if valor_saque >= 20:
      nota_20 = valor_saque // 20
      valor_saque %= 20

    if valor_saque >= 10:
      nota_10 = valor_saque // 10
      valor_saque %= 10

    if valor_saque >= 5:
      nota_5 = valor_saque // 5
      valor_saque %= 5

    if valor_saque >= 2:
      nota_2 = valor_saque // 2
      valor_saque %= 2

    if valor_saque >= 1:
      nota_1 = valor_saque // 1
      valor_saque %= 1 # Isso deve ser 0

    # Exibir apenas as notas utilizadas
    if nota_100 > 0:
      print(f"{nota_100} nota(s) de R$100")
    if nota_50 > 0:
      print(f"{nota_50} nota(s) de R$50")
    if nota_20 > 0:
      print(f"{nota_20} nota(s) de R$20")
    if nota_10 > 0:
      print(f"{nota_10} nota(s) de R$10")
    if nota_5 > 0:
      print(f"{nota_5} nota(s) de R$5")
    if nota_2 > 0:
      print(f"{nota_2} nota(s) de R$2")
    if nota_1 > 0:
      print(f"{nota_1} nota(s) de R$1")

except ValueError:
  print("Entrada inválida. Por favor, digite um valor inteiro para o saque.")
```
